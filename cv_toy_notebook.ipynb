## Scripted usage (no UI)

You can also use the underlying functions directly:

````python
from PIL import Image
import cv_toy

# Load an image
img = Image.open('my_lab_image.jpg')
cv_img = cv_toy.pil_to_cv(img)

# Register an object from a crop (x1,y1,x2,y2)
crop = cv_img[50:200, 80:300]
ok, msg = cv_toy.register_object('pipette_tip', crop)
print(msg)

# Find it in a new frame
annotated, msg = cv_toy.find_object_in_frame('pipette_tip', cv_img)
print(msg)
cv_toy.cv_to_pil(annotated)  # display inline in Jupyter
````

````python
import math

# Add points of interest and compute distance
pois = [(120, 80, 'tip_A'), (340, 200, 'tip_B')]
scale_mm_per_px = 0.05  # calibrate to your setup

annotated = cv_toy.draw_pois(cv_img, pois, scale_mm_per_px)

x1, y1, _ = pois[0]
x2, y2, _ = pois[1]
dist_px = math.sqrt((x2-x1)**2 + (y2-y1)**2)
print(f'Distance: {dist_px:.1f} px = {dist_px * scale_mm_per_px:.3f} mm')

cv_toy.cv_to_pil(annotated)
````
